```python
import requests
from bs4 import BeautifulSoup
def scrape_example():
    url = 'https://edunetfoundation.org/'
    response = requests.get(url)
    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')
        title = soup.title.get_text()
        print(f'title of the page:{title}')
    else:
        print(f"Failed to retrieve the page. Status code: {response.status_code}")
if __name__ == "__main__":
    scrape_example()
```

```python
#web crawling using Python
#import the requests Librabry
import requests

#detine the URL of the website to scrape
URL = "https://edunetfoundation.org/"

#send a GET request to the specitied URL and store the response in 'resp'
resp = requests.get(URL)

#print the HTTP status code of the response to check it the request was successful
print("Status Code:", resp.status_code)

#print the HIMI content of the response
print("\nResponse Content:")
print(resp.text)
```
